### Задача 4. Заказы в NumPy
Соберите из `rows` массивы `price`, `qty`, `product` (цикл для сбора списков можно, как на уроке). Посчитайте `revenue = price * qty`.
1. Сколько всего заказов?
2. Самый **дешёвый** заказ: сумма и товар (`argmin`).
3. Самый дорогой заказ: сумма. Во сколько раз он дороже самого дешёвого?

**Пример проверки:** `500` заказов; самый дешёвый — `7 500 ₸`, `Мышь Logitech M185`; самый дорогой — `1 098 000 ₸`.

````python
prices = []
for row in rows:
    price = int(row["price"].replace(" ₸", "").replace(" ", ""))
    prices.append(price)

quantities = []
for row in rows:
    quantity = int(row["qty"])
    quantities.append(quantity)

import numpy as np
revenue = np.array(prices) * np.array(quantities)
min_index = revenue.argmin()

print(f"{len(prices)} заказов")
print(f"самый дешевый заказ: {min(revenue)} ₸, {rows[min_index]["product"]}")
print(f"самый дорогой заказ: {max(revenue)} ₸")
print(f"{max(revenue) / min(revenue)} раз дороже")
````

````python
import json
with open("hw_base.json", "w", encoding="utf-8") as f:
    report = {
    "заказов": len(prices),
    "самый_дешёвый": {
        "сумма": int(min(revenue)),
        "товар": rows[min_index]["product"]
    },
    "самый_дорогой": int(max(revenue))
}
    json.dump(report, f, ensure_ascii=False, indent=4)

with open("hw_base.json", "r", encoding="utf-8") as f:
    data = json.load(f)
    print(data["самый_дешёвый"]["товар"])
````
